## 1. Import Matplotlib


```python
import matplotlib.pyplot as plt
```

## 4. Line Plot
Use a line plot to show trends or changes over an ordered/continuous variable.


```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Line Plot")
plt.grid(True)
plt.show()
```

## 5. Styling a Line Plot


```python
plt.plot(
    x, y,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=6
)
plt.show()
```

## 6. Multiple Lines on the Same Plot


```python
y2 = [1, 3, 5, 7, 9]

plt.plot(x, y, label="Line 1")
plt.plot(x, y2, label="Line 2")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Multiple Lines")
plt.legend()
plt.show()
```

## 7. Format Strings
Matplotlib format strings can combine marker, line style, and color.


```python
plt.plot(x, y, "o-b", label="Line 1")
plt.plot(x, y2, ">--g", label="Line 2")
plt.legend()
plt.show()
```

## 8. Movie Revenue Example
Compare Oscar-winning and non-Oscar-winning movies.


```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 839, 185, 611]

plt.plot(
    oscar_years,
    oscar_revenue,
    "o-b",
    label="Oscar Winners"
)

plt.plot(
    non_oscar_years,
    non_oscar_revenue,
    ">--g",
    label="Non-Oscar Winners"
)

plt.xlabel("Year")
plt.ylabel("Revenue ($M)")
plt.title("Movie Revenue")
plt.legend()
plt.grid(True)
plt.show()
```

## 9. Styles, Grid, Labels, Legend and Saving


```python
plt.style.use("ggplot")

plt.plot(oscar_years, oscar_revenue, label="Oscar Winners")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig("movie_revenue.png")
plt.show()

plt.style.use("default")
```

## 10. Bar Chart
Use bar charts to compare discrete categories.


```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.bar(movies, revenue)
plt.xlabel("Movies")
plt.ylabel("Revenue ($M)")
plt.title("Movie Revenue")
plt.xticks(rotation=30)
plt.show()
```

## 11. Styled Bar Chart


```python
plt.bar(
    movies,
    revenue,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5
)

plt.text(0, revenue[0], str(revenue[0]))
plt.xticks(rotation=30)
plt.show()
```

## 12. Horizontal Bar Chart


```python
plt.barh(movies, revenue)
plt.xlabel("Revenue ($M)")
plt.ylabel("Movies")
plt.title("Horizontal Bar Chart")
plt.show()
```

## 13. Grouped Bar Chart


```python
import numpy as np

x = np.arange(4)
y1 = [10, 15, 7, 12]
y2 = [8, 14, 9, 10]
width = 0.35

plt.bar(x - width / 2, y1, width, label="Group 1")
plt.bar(x + width / 2, y2, width, label="Group 2")

plt.xticks(x, ["A", "B", "C", "D"])
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Grouped Bar Chart")
plt.legend()
plt.show()
```

## 14. Scatter Plot
Use scatter plots to study the relationship between two numerical variables.


```python
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
salary = [30, 35, 42, 50, 58, 65, 72, 80, 88, 95]

plt.scatter(age, salary)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")
plt.grid(True)
plt.show()
```

## 15. Styled Scatter Plot


```python
plt.scatter(
    age,
    salary,
    s=100,
    c="blue",
    marker="o",
    alpha=0.7
)

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Styled Scatter Plot")
plt.show()
```

## 17. Pie Chart
Use pie charts for part-to-whole relationships with a small number of categories.


```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct="%.1f%%")
plt.title("Expenses")
plt.show()
```

## 18. Pie Chart — Explode, Shadow and Angle


```python
explode = [0, 0, 0, 0.2, 0]

plt.pie(
    amounts,
    labels=expenses,
    explode=explode,
    shadow=True,
    startangle=90,
    autopct="%.1f%%"
)
plt.title("Expenses")
plt.show()
```

## 19. Pie Chart — Wedge Properties


```python
plt.pie(
    amounts,
    labels=expenses,
    autopct="%.1f%%",
    wedgeprops={"edgecolor": "black", "linewidth": 2}
)
plt.show()
```

## 20. Histogram
Use a histogram to visualize the distribution of numerical data.


```python
np.random.seed(0)

scores = np.random.normal(
    loc=70,
    scale=10,
    size=100
)

plt.hist(
    scores,
    bins=10,
    edgecolor="black"
)

plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Score Distribution")
plt.show()
```

## 21. Histogram — Custom Bins and Orientation


```python
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

bins = [0, 30, 70, 90, 100]

plt.hist(marks, bins=bins, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()

plt.hist(marks, bins=bins, orientation="horizontal", edgecolor="black")
plt.show()
```

## 22. Multiple Histograms

## 24. Box Plot
A box plot summarizes a distribution using the minimum, Q1, median, Q3, maximum and possible outliers.


```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.ylabel("Values")
plt.title("Box Plot")
plt.grid(True)
plt.show()
```

## 25. Box Plot — Multiple Datasets


```python
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot(
    [group1, group2],
    tick_labels=["Group 1", "Group 2"]
)
plt.show()
```

## 26. Horizontal Box Plot and Mean


```python
plt.boxplot(
    data,
    vert=False,
    showmeans=True,
    whis=2.0
)
plt.xlabel("Values")
plt.grid(True)
plt.show()
```

## 27. Stack Plot
Use stack plots to show how multiple components contribute to a total over time.


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.stackplot(
    days,
    direct,
    organic,
    social,
    labels=["Direct", "Organic", "Social"]
)

plt.xlabel("Day")
plt.ylabel("Visits")
plt.title("Traffic Sources")
plt.legend(loc="upper left")
plt.show()
```

## 28. Subplots
Display multiple plots inside one figure.


```python
x = list(range(1, 11))

y1 = [np.sqrt(i) for i in x]
y2 = [i * 2 for i in x]
y3 = [i ** 2 for i in x]
y4 = [i ** 3 for i in x]

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title("Square Root")

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title("Double")

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title("Square")

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title("Cube")

plt.tight_layout()
plt.show()
```

## 29. Modern Matplotlib — Object-Oriented API
Use `fig` and `ax` objects for more control over figures and multiple axes.


```python
fig, ax = plt.subplots()

vals = [1, 2, 3, 4, 5, 6]

ax.plot(vals, vals)
ax.set(
    title="Sample Plot",
    xlabel="X axis values",
    ylabel="Y axis values"
)
ax.grid(True)

fig.tight_layout()
plt.show()
```

## 30. Object-Oriented Subplots


```python
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].plot(x, y1)
axes[0, 0].set_title("Top Left")

axes[0, 1].bar(["A", "B", "C"], [3, 5, 2])
axes[0, 1].set_title("Top Right")

axes[1, 0].scatter(x, y2)
axes[1, 0].set_title("Bottom Left")

axes[1, 1].hist(marks, bins=5)
axes[1, 1].set_title("Bottom Right")

fig.tight_layout()
plt.show()
```

## 31. Practical Example — Temperature Over Cities


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [20, 21, 22, 23, 24]
]

for city, temps in zip(cities, temperatures):
    plt.plot(days, temps, marker="o", label=city)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Over Cities")
plt.legend()
plt.grid(True)
plt.show()
```
