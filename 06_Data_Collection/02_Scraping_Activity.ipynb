# Web Scraping Activity

## Activity
Scrape quotes from all available pages of Quotes to Scrape, save each page as an HTML file, then extract quotes tagged with **life** from page 1.

```python
import requests
from bs4 import BeautifulSoup
```

```python
# Fetch data & store it

page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    
    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore..")
        break

    with open(f"scraped_data/quotes{page_count}.html", "w") as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")

    page_count = page_count + 1
```

```python
# Extract useful info

life_quotes = []

for page in range(1, page_count):
    with open(f"scraped_data/quotes{page}.html", "r") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append(
                {
                    "quote": text,
                    "author": author,
                    "page": page,
                }
            )

print(life_quotes)
```

```python
import pandas as pd

df = pd.DataFrame(life_quotes)

df.to_csv("cleaned_data/life_quotes.csv", index=False)

print("Data saved to cleaned_data/life_quotes.csv")
```
