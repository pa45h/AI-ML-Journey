# 1. What is Pandas?

**Pandas** is a Python library for data analysis and manipulation. It provides high-level structures for labeled and tabular data and handles heterogeneous data and missing values conveniently.

Its two core structures are:

- **Series** — one-dimensional labeled array
- **DataFrame** — two-dimensional labeled table

Pandas is heavily used together with NumPy, Matplotlib and Scikit-learn.

````python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
````

# 2. Series

A Series contains:

1. **Values** — actual data
2. **Index** — labels for each value

A Series supports vectorized operations and can represent missing values.

````python
s = pd.Series([23, 24, 25, 26])

print(s)
print(type(s))
print("Index:", s.index)
print("Values:", s.values)

print("\nPosition 0:", s.iloc[0])
print("Position 2:", s.iloc[2])
````

### Custom indexing

````python
s = pd.Series(
    [23, 24, 25, 26],
    index=["Adam", "Eve", "Charlie", "Bob"]
)

print(s)
print("\nEve:", s["Eve"])
print("Bob:", s["Bob"])
````

### Series operations and label alignment

Pandas aligns Series arithmetic by their **index labels**, not simply by physical position.

````python
s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])
print(s1 + s2)

a = pd.Series([10, 20, 30], index=["a", "b", "c"])
b = pd.Series([1, 2, 3], index=["b", "c", "d"])

print("\nLabel-aligned result:")
print(a + b)
````

# 11. Datetime Data

Convert timestamp strings using:

```python
pd.to_datetime(...)
```

Then use `.dt` to extract components such as date, year, month and hour.

````python
time_df = pd.DataFrame({
    "timestamp": [
        "2025-11-04 18:25:17",
        "2025-11-05 09:10:00",
        "2025-12-10 14:30:00"
    ],
    "aqi":[120,80,160]
})

time_df["timestamp"] = pd.to_datetime(time_df["timestamp"])
time_df["date"] = time_df["timestamp"].dt.date
time_df["year"] = time_df["timestamp"].dt.year
time_df["month"] = time_df["timestamp"].dt.month
time_df["hour"] = time_df["timestamp"].dt.hour

print(time_df)
````

# 12. Missing Values

Important methods:

- `isnull()` / `isna()`
- `dropna()`
- `fillna()`
- `ffill()`
- `bfill()`

````python
missing = pd.DataFrame({
    "name":["Alice","Bob","Charlie","Diana"],
    "age":[25,np.nan,30,np.nan],
    "income":[50000,60000,np.nan,80000]
})

print("Missing mask:\n", missing.isnull())
print("\nMissing count:\n", missing.isnull().sum())
````

````python
print("Drop rows:\n", missing.dropna())
print("\nDrop columns:\n", missing.dropna(axis=1))

filled = missing.copy()
filled["age"] = filled["age"].fillna(filled["age"].mean())
filled["income"] = filled["income"].fillna(filled["income"].median())

print("\nFilled:\n", filled)
````

````python
s = pd.Series([10, np.nan, np.nan, 40, np.nan])

print("Original:\n", s)
print("\nForward fill:\n", s.ffill())
print("\nBackward fill:\n", s.bfill())
````

# 16. Transformation / Feature Engineering

Course methods:

- `apply()` — apply a function
- `map()` — map Series values
- `assign()` — create/modify columns
- `replace()` — replace values

````python
feature = pd.DataFrame({
    "income":[40000,55000,75000,100000],
    "gender":["Male","Female","Unknown","Female"],
    "country":["USA","India","USA","India"]
})

feature["tax_rate"] = feature["income"].apply(
    lambda x: "10%" if x < 60000 else "20%"
)

feature["gender"] = feature["gender"].map(
    {"Male":"M","Female":"F","Unknown":"U"}
)

feature = feature.assign(
    income_after_raise=feature["income"] * 1.10
)

feature["country"] = feature["country"].replace({
    "USA":"United States"
})

print(feature)
````

### Vectorization vs `apply()`

For simple numerical operations, prefer vectorized operations:

```python
df["income"] * 1.10
```

instead of:

```python
df["income"].apply(lambda x: x * 1.10)
```

`apply()` remains useful when the operation cannot be expressed conveniently with built-in vectorized operations.

````python
x = pd.DataFrame({"income":[50000,60000,70000]})
x["new_income"] = x["income"] * 1.10
print(x)
````

# 20. GroupBy and Aggregation

`groupby()` splits data into groups so that aggregations can be calculated per group.

Common aggregations:
- `sum`
- `mean`
- `count`
- `min`
- `max`
- `std`

````python
g = pd.DataFrame({
    "country":["India","India","USA","USA","India"],
    "gender":["M","F","M","F","M"],
    "age":[25,30,35,28,40],
    "income":[50000,60000,80000,70000,90000]
})

print(g.groupby("country")["income"].mean())
print("\nGender mean:")
print(g.groupby("gender")["income"].mean())

print("\nCountry income aggregates:")
print(g.groupby("country")["income"].agg(["mean","min","max"]))
````

````python
print(
    g.groupby("country").agg(
        avg_age=("age","mean"),
        avg_income=("income","mean"),
        max_income=("income","max")
    )
)

print("\nMultiple grouping columns:")
print(g.groupby(["country","gender"])["income"].mean())
````

# 21. Melt — Wide to Long

`melt()` converts wide-form data into long-form data.

Key parameters:
- `id_vars`
- `value_vars`
- `var_name`
- `value_name`

````python
sales = pd.DataFrame({
    "country":["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22]
})

melted = sales.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metric",
    value_name="value"
)

print(melted)
````

# 22. Pivot — Long to Wide

`pivot()` reshapes long data into wide data.

If combinations are duplicated, `pivot()` cannot resolve which value should occupy a cell; `pivot_table()` is the aggregation-aware alternative.

````python
pivoted = melted.pivot(
    index=["country","year"],
    columns="metric",
    values="value"
)

print(pivoted)
````

# 23. Merge and SQL-Style Joins

`pd.merge()` combines DataFrames using common keys.

Join types:
- inner
- left
- right
- outer

````python
customers = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})

orders = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})

for how in ["inner","left","right","outer"]:
    print(f"\n{how.upper()} JOIN:")
    print(pd.merge(customers, orders, on="customer_id", how=how))
````

### Merge on differently named keys

Use `left_on` and `right_on` when the key columns have different names.

````python
left = pd.DataFrame({"customer_id":[1,2,3], "name":["A","B","C"]})
right = pd.DataFrame({"client_id":[1,2,4], "amount":[100,200,300]})

print(pd.merge(
    left, right,
    left_on="customer_id",
    right_on="client_id",
    how="left"
))
````

# 32. `transform()` — Group-Level Features

`agg()` reduces groups.

`transform()` returns values aligned with the original rows.

This makes `transform()` particularly useful for feature engineering.

````python
sales2 = pd.DataFrame({
    "country":["India","India","USA","USA"],
    "income":[50,70,60,90]
})

sales2["country_mean"] = (
    sales2.groupby("country")["income"].transform("mean")
)

sales2["difference_from_country_mean"] = (
    sales2["income"] - sales2["country_mean"]
)

print(sales2)
````

# 34. Merge Validation

The `validate` parameter can catch unexpected relationship types in joins.

For example, customer → orders is normally a **many-to-one** relationship when merging orders with a unique customer table.

````python
customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "name":["A","B","C"]
})

orders = pd.DataFrame({
    "order_id":[101,102,103],
    "customer_id":[1,1,2],
    "amount":[100,200,300]
})

merged = pd.merge(
    orders, customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

print(merged)
````

# 42. Avoiding Chained Assignment


# 48. Standardization with Pandas

NumPy introduced standardization earlier:

\[
z = \frac{x - \mu}{\sigma}
\]

For a DataFrame, calculating column means and standard deviations can demonstrate feature-wise standardization.

For actual ML training, fit preprocessing only on the training set to avoid leakage.

````python
features = pd.DataFrame({
    "height":[160,170,180,190],
    "income":[40000,50000,60000,80000]
})

standardized = (
    features - features.mean()
) / features.std()

print(standardized)
````

# 49. Data Leakage — Critical ML Concept

A preprocessing statistic must not learn from the test set.

Incorrect idea:

```text
Full dataset
   ↓
calculate mean/std
   ↓
split train/test
```

Safer workflow:

```text
Raw dataset
   ↓
Train/test split
   ↓
Fit preprocessing on training data
   ↓
Transform train + test using training parameters
```

Pandas is excellent for data preparation, but production ML preprocessing should eventually use tools such as Scikit-learn pipelines to enforce this workflow.

# 50. Practical End-to-End Cleaning Pipeline

A typical workflow:

```text
Load
 ↓
Inspect
 ↓
Clean column names
 ↓
Convert types
 ↓
Handle missing values
 ↓
Remove duplicates
 ↓
Clean strings
 ↓
Filter invalid records
 ↓
Create features
 ↓
Aggregate / merge / reshape
 ↓
Validate
 ↓
Export
```

````python
raw = pd.DataFrame({
    " Name ":[" Alice ","Bob","Bob","Charlie"],
    "Age":["25","30","30","bad"],
    "Income":["50000","65000","65000","70000"],
    "Country":["india","USA","USA","India"]
})

clean = raw.copy()

clean.columns = clean.columns.str.strip().str.lower()
clean["name"] = clean["name"].str.strip().str.title()
clean["country"] = clean["country"].str.strip().str.title()

clean["age"] = pd.to_numeric(clean["age"], errors="coerce")
clean["income"] = pd.to_numeric(clean["income"], errors="coerce")

clean = clean.drop_duplicates()
clean = clean.dropna(subset=["age","income"])

clean["income_k"] = clean["income"] / 1000

print(clean)
````

5. Forgetting parentheses around Boolean conditions
6. Ignoring index alignment
7. Treating numeric strings as numbers
8. Blindly filling missing values
9. Using `apply()` where vectorization is available
10. Modifying filtered data ambiguously
