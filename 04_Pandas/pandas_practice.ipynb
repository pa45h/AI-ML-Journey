# 1. Series Practice

A **Series** is a one-dimensional labeled data structure.

A Series contains:
- values
- index labels

Let's create one and inspect its basic properties.

````python
s = pd.Series([23, 24, 25, 26])

print(s)
print(type(s))

print("\nIndex labels:")
print(s.index)

print("\nValues:")
print(s.values)
````

### Accessing Series values

With the default integer index, positions such as `0` and `2` can be used to access values.

`iloc` is the explicit position-based form.

````python
print("First value:", s[0])
print("Third value:", s[2])

print("\nUsing iloc:")
print("First value:", s.iloc[0])
print("Third value:", s.iloc[2])
````

### Custom indexing

A Series can use meaningful labels instead of the default integer index.

````python
s2 = pd.Series(
    [23, 24, 25, 26],
    index=["Adam", "Eve", "Charlie", "Bob"]
)

print(s2)

print("\nEve:", s2["Eve"])
print("Bob:", s2["Bob"])
````

### Vectorized operations

Pandas Series support vectorized operations, allowing arithmetic to be applied to complete Series objects.

````python
s1 = pd.Series([1, 2, 3])
s2 = pd.Series([4, 5, 6])

print("s1 + s2:")
print(s1 + s2)

print("\ns1 * 10:")
print(s1 * 10)
````

### Mutable values

Individual values can be changed.

However, the tutorial demonstrates treating the Series size as fixed for normal use; operations such as `drop()` return a modified result rather than changing the original object's structure in place by default.

````python
s = pd.Series([1, 2, 3, 4, 5])

s[0] = 100

print("After changing the first value:")
print(s)

changed_s = s.drop(1)

print("\nAfter drop(1):")
print(changed_s)

print("\nOriginal Series is unchanged:")
print(s)
````

### Multiple conditions

Use:

- `|` for OR

Put each condition in parentheses.

````python
target_timestamp = "2025-11-04 18:25:17.554219"

aqi_data = air[
    (air["aqi"] > 100) &
    (air["timestamp"] == target_timestamp)
][["city", "aqi"]]

print(aqi_data)
````

# 11. `.loc` vs `.iloc` on a Filtered DataFrame

A filtered DataFrame can retain its original index.

Therefore:

```python
aqi_data.loc[360]
```

means **index label 360**, while:

```python
aqi_data.iloc[1]
```

means **the second row by position**.

They can refer to different rows even when both syntaxes look similar.

````python
if len(aqi_data) > 1:
    print("Second row by position:")
    print(aqi_data.iloc[1])

    print("\nIndex labels present:")
    print(aqi_data.index)
````

# 33. Melt — Wide to Long

`melt()` converts a wide DataFrame into a long-form DataFrame.

We keep:
- `country`
- `year`

and unpivot:
- `sales`
- `profit`

````python
sales_df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

melted = sales_df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metric",
    value_name="value (in Lakhs)"
)

print(melted)
````

# 35. Merging DataFrames

We have two related datasets:

- customer information
- order information

Both contain `customer_id`, which can be used as the common key.

````python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

print("Customers:")
print(df_customers)

print("\nOrders:")
print(df_orders)
````

## Inner Join

Keeps only matching keys from both DataFrames.

````python
inner_join = pd.merge(
    df_customers,
    df_orders,
    on="customer_id"
)

print(inner_join)
````

## Left Join

Keeps every row from the left DataFrame and adds matching data from the right DataFrame.

````python
left_join = pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="left"
)

print(left_join)
````

## Right Join

````python
right_join = pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="right"
)

print(right_join)
````

## Outer Join

Keeps keys from both DataFrames.

````python
outer_join = pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="outer"
)

print(outer_join)
````

# 38. Practice Workflow — Clean `raw_data.csv`

Now combine several concepts into one practical workflow.

Goal:
1. Load the raw dataset.
2. Inspect it.
3. Remove duplicates.
4. Handle missing values.
5. Convert numeric data.
6. Clean strings.
7. Create a feature.
8. Sort the result.

````python
raw = pd.read_csv("raw_data.csv")

print("Original shape:", raw.shape)
print("\nMissing values:")
print(raw.isna().sum())
````

````python
clean = raw.copy()

# Remove exact duplicate rows
clean = clean.drop_duplicates()

# Convert income safely
clean["income"] = pd.to_numeric(
    clean["income"],
    errors="coerce"
)

# Convert age safely
clean["age"] = pd.to_numeric(
    clean["age"],
    errors="coerce"
)

# Clean string columns
clean["name"] = clean["name"].str.strip()
clean["country"] = clean["country"].str.strip()
clean["gender"] = clean["gender"].str.strip()

# Remove rows where required numerical fields are missing
clean = clean.dropna(
    subset=["age", "income"]
)

# Create a simple feature
clean["income_k"] = clean["income"] / 1000

# Sort
clean = clean.sort_values(
    "income",
    ascending=False
)

print(clean)
````
