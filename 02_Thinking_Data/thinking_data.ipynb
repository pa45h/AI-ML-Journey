```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("data.json")
print(data)
```

```python
def clean_data(data):
    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}

    existing_users = set()
    cleaned_data = []

    for user in data:
        
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        if(user["name"].strip() in existing_users):
            continue
        existing_users.add(user["name"].strip())
        cleaned_data.append(user)

    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
def get_insights(data):

    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])

    print(f"average product rating = {total_rating/len(data)}")

    poor_rating = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating += 1

    print(f"% of poor rating = {poor_rating/len(data)*100}%")
```

```python
get_insights(data)
```

```python
def get_recommendations(data):
    
    recommendations = []
    for user in data:
        
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        
        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendations(data)
```
